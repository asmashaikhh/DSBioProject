## 1. Introduction

The goal of the research paper "The effects of temperature on the defensive strikes of rattlesnakes" was to see if the body temperature rattlesnakes would affect how they strike predators. The researchers placed the snakes in a temperature controlled room overnight then collected information on their behaviors and kinematics while they striked an aggressively moving balloon that acted as a predator.

### 2.1
**Section 1:**

a) i. 	To create the data, 24 snakes were placed in a test room for 12 hours to acclimate to the temperature. Then, they were placed in an enclosure set to a specific temperature. ### 3.3 Visualizations

Graph 1- velocity of snake (measured by their necks in m/s) vs skin temperature
For this graph, I will make a scatter plot that shows the velocity of the snake and their skin temperatures to see if skin temperature shows some relation to the snakes' velocities. This is appropriate for this feature and data set as scatter plots can show if there is a trend in data based on where the individual points lie. This visualization will allow viewers to see if there is a change in velocity as temperature changes. 

```python
#make a scatterplot through seaborn using the cleaned up data frame where the x axis shows the skin temperature and the y axis shows the velocities of the snakes' necks
sns.relplot(data=snakedf, x='SKIN_TEMP', y='NECK_MAX_VEL')
```

This graph mostly shows that as skin temperature increases, the snakes reach a higher maximum velocity. Using the data from the describe along with this graph, it can be seen that the minimum velocity of 1.52m/s is below 15°C, and the highest velocity of 5.3m/s is just below 30°C. The data is a bit spread out which suggests a smaller correlation, however I do think there is a general trend of the data as a hypothetical line could be made with the data points relatively close to the line. This helps us see that there is a a slight trend between the two and that the snakes' temperature likely affects their velocities. This relates to the hypothesis of the paper as shows the velocity, a feature of the snakes' strike being affected by the body temperature, and specifically that the performances is decreasing at lower body temperatures. 

Graph 2- Strike Distance vs Skin Temperature

For this graph, I will make a barplot that compares the 5 temperature treatments to the snakes' strike distances so we will see the averages of the snakes' strike distances in each temperature treatment. A bar plot is best to use in this case as it shows the averages for each category of temperature and will make it easier to visualize if there is a difference between the different temperatures. 

```python
#make a barplot using seaborn and the cleaned up dataframe where the x axis shows the different temperature conditions and the y axis shows the strike distances.
sns.barplot(data=snakedf, x='TEMP_TREATMENT', y='STRIKE_DISTANCE')
```

This graph shows that at higher temperatures, the snakes seem to strike from a further distance. This further distance can show that the snake is more prepared for an attack. Therefore, a longer strike distance is better. If the snake is too close, then it may show that the snake did not have quick reflexes to react to the balloon. While treatment temperatures between 20-30°C do not show much of a difference, with 25°C being slightly higher than both, there is a clear difference between 15°C and 35°C, therefore I made the conclusion that the graph does show a trend. In relation to the describe function and the feature, the maximum value of 35.5cm can be seen at the higher temperatures and while the minimum strike distance is not explicitly shown, the bar with the lowest strike distances is also the bar with the lowest temperature. The mean can also be seen in the graph as most of the data lies around 16cm. This graph relates to the hypothesis and is consistent with the other figures in the paper as they all show that strike performance is dependent on temperature of the snakes.
